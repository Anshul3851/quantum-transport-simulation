# Stage 3: disorder and localization in coherent quantum transport

This notebook inspects the reproducible finite-device disorder study. It loads saved data generated by `scripts/run_transport_stage3.py`; it does not rerun the ensemble. The onsite convention is `epsilon_i = epsilon_0 + w_i`, with independent `w_i ~ Uniform(-W/2, W/2)`. Transmission is unchanged by any log-floor operation; the floor is used only for `ln(T)`.

```python
from pathlib import Path
import csv
import json
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "results/stage3_ensemble.csv").exists():
    ROOT = ROOT.parent
metadata = json.loads((ROOT / "results/stage3_metadata.json").read_text())
def read_csv(name):
    with (ROOT / "results" / name).open(newline="") as handle:
        return list(csv.DictReader(handle))
ensemble = read_csv("stage3_ensemble.csv")
lengths = read_csv("stage3_length_scaling.csv")
print(metadata["ensemble"])
print(ensemble[0])
```

## Single-realization spectra

The panel below overlays four independently sampled onsite profiles for each W. Individual curves are shown to expose realization-to-realization variation; they are not ensemble averages.

```python
fig, ax = plt.subplots()
ax.imshow(plt.imread(ROOT / "figures/stage3_disorder_realizations.png"))
ax.axis("off")
plt.show()
```

## N=40 ensemble

The table shows the saved summaries for `E = 0, 0.5, 1`. Standard deviation uses `ddof=1`, SEM is SD divided by `sqrt(n)`, and `Ttyp = exp(mean(log(T)))`. The stored floor count is zero for these samples.

```python
for row in ensemble:
    print(*(row[key] for key in ("W", "energy", "n_realizations", "mean_T", "median_T",
                                  "standard_deviation_T", "standard_error_mean_T",
                                  "mean_log_T", "typical_T", "log_floor_count")), sep="\t")
```

```python
for filename in ("stage3_transmission_disorder.png", "stage3_ensemble_statistics.png"):
    fig, ax = plt.subplots()
    ax.imshow(plt.imread(ROOT / "figures" / filename))
    ax.axis("off")
    plt.show()
```

## Length dependence

The same 100 disorder profiles for each `(N, W)` are evaluated at E=0 and E=0.5. The table and figure describe finite-size trends. Any fitted slope is exploratory and is not a thermodynamic localization-length result.

```python
for row in sorted(lengths, key=lambda r: (float(r["W"]), float(r["energy"]), int(r["N"]))):
    print(*(row[key] for key in ("N", "W", "energy", "mean_T", "mean_log_T", "typical_T",
                                  "standard_error_mean_T", "log_floor_count")), sep="\t")
```
